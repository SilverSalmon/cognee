# Lane Initial Checker

A tracer-bullet test that sends a single relationship across ADF, DBW, and Unity Catalog into Cognee, builds the knowledge graph, and verifies the result in the UI.

```python
import asyncio
import cognee

async def test_run():
    # This is the "Tracer Bullet" - a single, clear path for Cognee to map.
    # It bridges the gap between your three repos: ADF, DBW, and UC.
    data = (
        "The Schedule_Daily_6AM trigger starts the Ingest_Sales_ADF pipeline, "
        "which runs the Transform_Sales_DBW job to update the Gold_Finance_Table "
        "in Unity Catalog."
    )

    print("\U0001f9e0 Sending relationship to Cognee...")
    # 'remember' puts the data into the staging area
    await cognee.remember(data)

    print("\u2699\ufe0f  Building the Knowledge Graph...")
    # 'cognify' is the heavy lifting where the LLM extracts the nodes and edges
    await cognee.cognify()

    print("\n\u2705 Success! Your server logic is now mapped.")
    print("You can now check the UI at http://localhost:3000 to see the nodes.")

await test_run()
```

```python
import os
from pathlib import Path
import cognee

REPO_PATH = Path(r"C:\dev\dataforge\dataforge-us-dbw")

# Directories to skip (noise, backups, deprecated)
IGNORE_DIRS = {".localwork", ".git", "__pycache__", ".ipynb_checkpoints", "deprecated"}

# File extensions worth cognifying
INCLUDE_EXTENSIONS = {".py", ".ipynb", ".yml", ".yaml", ".json", ".md"}

# Files to skip by name
IGNORE_FILES = {".gitignore", "package.json", "package-lock.json"}

# Max file size to ingest (skip huge generated files)
MAX_FILE_SIZE = 100_000  # 100KB


def should_include(path: Path) -> bool:
    """Filter for relevant DevOps/Databricks files."""
    # Skip ignored directories
    for part in path.parts:
        if part in IGNORE_DIRS:
            return False

    # Skip by filename
    if path.name in IGNORE_FILES:
        return False

    # Only include relevant extensions
    if path.suffix.lower() not in INCLUDE_EXTENSIONS:
        return False

    # Skip empty or oversized files
    try:
        size = path.stat().st_size
        if size == 0 or size > MAX_FILE_SIZE:
            return False
    except OSError:
        return False

    return True


def read_file_content(path: Path) -> str | None:
    """Read file content, handling notebooks specially."""
    try:
        if path.suffix == ".ipynb":
            import json
            with open(path, "r", encoding="utf-8") as f:
                nb = json.load(f)
            # Extract only source code from cells (skip outputs/metadata noise)
            cells = nb.get("cells", [])
            sources = []
            for cell in cells:
                cell_type = cell.get("cell_type", "")
                source = "".join(cell.get("source", []))
                if source.strip():
                    sources.append(f"# [{cell_type}]\n{source}")
            return "\n\n".join(sources) if sources else None
        else:
            with open(path, "r", encoding="utf-8", errors="ignore") as f:
                return f.read()
    except Exception as e:
        print(f"  ⚠️ Could not read {path.name}: {e}")
        return None


async def cognify_dbw_repo():
    """Cognify the dataforge-us-dbw repo with filtering."""
    files = [f for f in REPO_PATH.rglob("*") if f.is_file() and should_include(f)]

    print(f"📂 Repo: {REPO_PATH}")
    print(f"✅ Found {len(files)} relevant files:\n")
    for f in sorted(files):
        print(f"  {f.relative_to(REPO_PATH)}")

    print(f"\n🧠 Adding {len(files)} files to Cognee...")

    for fpath in sorted(files):
        content = read_file_content(fpath)
        if not content or not content.strip():
            continue

        rel_path = fpath.relative_to(REPO_PATH)
        # Context header — avoid starting with "File:" as urlparse treats it as a scheme
        context = f"[Source: {rel_path} | Repo: dataforge-us-dbw | Type: {fpath.suffix}]"
        text = f"{context}\n\n{content}"

        await cognee.add(text, dataset_name="dataforge_dbw")
        print(f"  ✓ {rel_path}")

    print("\n⚙️  Building Knowledge Graph...")
    await cognee.cognify(datasets=["dataforge_dbw"])
    print("✅ Done! Dataset: dataforge_dbw")


await cognify_dbw_repo()
```
