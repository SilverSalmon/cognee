## Search the graph

After cognify completes, run a search to verify data made it into the knowledge graph.

```python
import cognee
from cognee.modules.search.types import SearchType

results = await cognee.search(
    query_text="What rules and standards exist?",
    query_type=SearchType.GRAPH_COMPLETION,
)

for r in results:
    print(r)
```

```python
# Diagnostic: check what's actually in the graph
from cognee.infrastructure.databases.graph import get_graph_engine

graph_engine = await get_graph_engine()

# Count nodes and edges
nodes = await graph_engine.get_graph_data()
print(f"Graph data type: {type(nodes)}")
print(f"Graph data keys: {nodes.keys() if isinstance(nodes, dict) else 'N/A'}")
if isinstance(nodes, dict):
    for k, v in nodes.items():
        print(f"  {k}: {len(v) if hasattr(v, '__len__') else v}")

# Also check what results actually is
print(f"\nSearch results type: {type(results)}")
print(f"Search results length: {len(results)}")
print(f"Search results value: {results!r}")
```
